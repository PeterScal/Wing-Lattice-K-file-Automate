```python
import sys

# import for DYNA
from ansys.dyna.core import keywords as kwd
from ansys.dyna.core.pre.dynabase import (
    SolidPart,
    DynaBase,
    Part,
    
)
from ansys.dyna.core.pre.dynamaterial import (
    MatRigid,
    MatPiecewiseLinearPlasticity,
    MatModifiedPiecewiseLinearPlasticity,
)

from ansys.dyna.core.pre import launch_dynapre
from ansys.dyna.core.pre.dynamech import (
    AnalysisType,
    DynaMech,
    SolidPart,
    Part,
    SolidFormulation
)
from ansys.dyna.core.pre.misc import check_valid_ip



mat_ABS = MatPiecewiseLinearPlasticity(
    mass_density=1.25e-9, 
    young_modulus=1.0, 
    yield_stress=0.05, 
    poisson_ratio=0.35,
    tangent_modulus=0.02
)



Solid = SolidPart(1)
Solid.set_element_formulation(SolidFormulation.CONSTANT_STRESS_SOLID_ELEMENT)
# set_material expects the material object (or named args); pass the material first
Solid.set_material(mat_ABS)
dynamech.parts.add(Solid)



```
